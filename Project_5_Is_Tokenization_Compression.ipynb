# Part_1:Text compression in English Books

## Task_1:
- In a for-loop over books, count the number of characters, the number
of words (using the .split() method), and the number of tokens in
each book. Calculate two compression ratios: the number of tokens
divided by the number of characters, and the number of tokens divided by the number of words. Print the information in a table as
below.

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained('gpt2')
```

```python
baseurl = 'https://www.gutenberg.org/cache/epub/'

bookurls = [
    # code       title
    ['84',    'Frankenstein'    ],
    ['64317', 'GreatGatsby'     ],
    ['11',    'AliceWonderland' ],
    ['1513',  'RomeoJuliet'     ],
    ['76',    'HuckFinn'        ],
    ['219',   'HeartDarkness'   ],
    ['2591',  'GrimmsTales'     ],
    ['2148',  'EdgarAllenPoe'   ],
    ['36',    'WarOfTheWorlds'  ],
    ['829',   'GulliversTravels']
]
```

```python
print('  Book title     |  Chars  |  Words  |  Tokens |  Compress')
print('-----------------+---------+---------+---------+-----------')

compression = np.zeros((len(bookurls),2))
i = 0

for code,title in bookurls:

  # get the text
  fullurl = baseurl + code + '/pg' + code + '.txt'
  text = requests.get(fullurl).text

  # counts
  num_chars  = len(text)
  num_words  = len(text.split())
  num_tokens = len(tokenizer.encode(text))

  # compression ratios
  compression[i,0] = 100*num_tokens/num_chars
  compression[i,1] = 100*num_tokens/num_words

  print(f'{title:16} | {num_chars:>7,d} | {num_words:>7,d} | {num_tokens:>7,d} | {compression[i,0]:>3.0f} / {compression[i,1]:>3.0f}')
  i += 1
```
