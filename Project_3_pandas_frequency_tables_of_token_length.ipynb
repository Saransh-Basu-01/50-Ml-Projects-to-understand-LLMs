```python
from transformers import AutoTokenizer
tokenizer=AutoTokenizer.from_pretrained('gpt2')
```

```python
txt = "Pulcinella is a 21-section ballet by Igor Stravinsky with arias for soprano, tenor and bass vocal soloists, and two sung trios. It is based on the 18th-century play Quatre Polichinelles semblables, or Four similar Pulcinellas, revolving around a stock character from commedia dell'arte. The work premiered at the Paris Opera on 15 May 1920 under the baton of Ernest Ansermet. The central dancer, Léonide Massine, created both the libretto and the choreography, while Pablo Picasso designed the costumes and sets. The ballet was commissioned by Sergei Diaghilev, impresario of the Ballets Russes. A complete performance takes 35–40 minutes. Stravinsky revised the score in 1965."
```

```python
tokens=tokenizer.encode(txt)
```

```python
decode_text=tokenizer.decode(tokens)
```

## Task 2
- Create a pandas dataframe that contains 174 rows (one per token)
and two columns. The first column is the token index and the second
column is the number of characters in that token. Print out the first
five rows.
Use the dataframe method .value_counts() to create a bar
plot of token length distributions, as in Figure 2.8. Use the Seaborn
library. I used the color palette plasma

```python
token_lens = [len(tokenizer.decode(t)) for t in tokens]
```

```python
df=pd.DataFrame({'tokens_index':tokens,'token_length':token_lens})
```

```python
df.head()
```

```python
df['token_length'].value_counts()
```

```python

val_counts=df['token_length'].value_counts()
```

```python
# specify figure size
plt.figure(figsize=(10,4))

# draw the bar plot
sns.barplot(x=val_counts.index,y=val_counts.values,edgecolor='k',
            hue=val_counts.values,palette='plasma',legend=False)

# adjustments
plt.gca().set(xlabel='Token length (characters)',ylabel='Count')

plt.tight_layout()
plt.show()
```

# Part 2:Distribution of book token lengths

## Task 1:
- In a for-loop over the list of book titles and website codes from
Project 1, import and tokenize each book, create the dataframe and
Series as in Part 1 of this project, and visualize the results in a
scatter plot, using Seaborn’s scatterplot function, as in Figure 2.9.
Books have different lengths, so you will need to normalize the
counts.

```python
baseurl = 'https://www.gutenberg.org/cache/epub/'

bookurls = [
    # code       title
    ['84',    'Frankenstein'    ],
    ['64317', 'GreatGatsby'     ],
    ['11',    'AliceWonderland' ],
    ['1513',  'RomeoJuliet'     ],
    ['76',    'HuckFinn'        ],
    ['219',   'HeartDarkness'   ],
    ['2591',  'GrimmsTales'     ],
    ['2148',  'EdgarAllenPoe'   ],
    ['36',    'WarOfTheWorlds'  ],
    ['829',   'GulliversTravels']
]
```

```python
import requests
```

```python
plt.figure(figsize=(10,4))

markersymbols = 'soh^vP*DXp'

for i,(code,title) in enumerate(bookurls):
  fullurl = baseurl + code + '/pg' + code + '.txt'
  tokens = tokenizer.encode(requests.get(fullurl).text)
  tokens_len=[len(tokenizer.decode(x)) for x in tokens]
  df=pd.DataFrame({'tokens_index':tokens,'token_length':tokens_len})
  val_counts=df['token_length'].value_counts()
  sns.scatterplot(x = val_counts.index,
                  y = val_counts.values,
                  s = 80,
                  alpha  = .5,
                  marker = markersymbols[i],
                  edgecolor = 'k',
                  label  = title)
plt.gca().set(xlabel='Token length (characters)',ylabel='Frequency (norm.)',xscale='log')

plt.tight_layout()
plt.show()
```
