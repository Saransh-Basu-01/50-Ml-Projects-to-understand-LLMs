# Part 1:Character Based Tokenization

```python
txt ='The way you do anything is the way you do everything.'
```

```python
characters=[c for c in txt]
```

```python
len(characters)
```

```python
char_vocab=list(sorted(set(characters)))
```

```python
len(char_vocab)
```

## Task 1
- Write code to create a look-up-table that maps each token (a string)
onto an integer.

```python
char_vocab
```

```python
print('Token :',end='')
for v in char_vocab:
  print(f' "{v}" | ', end='')
print('\n','-'*131)

print('Index :',end='')
[print(f' {i:2}  | ',end='') for i in range(len(char_vocab))];
```

## Task 2
- Write code to tokenize (encode) the text by transforming it from a
list of characters into a sequence of integers, which you can print out.



```python

tokens = [0]*len(txt)

for i,c in enumerate(txt):
  tokens[i] = char_vocab.index(c)

print(tokens)
```

```python
char_vocab.index('a')
```

## Task 3
- Visualize the token sequence in a plot like in Figure 2.3. The x-axis
corresponds to positional indices, and the y-axis corresponds to the
token indices on the left and the token strings on the right. Draw a
box at the y-axis location of each token index

```python
import matplotlib.pyplot as plt
_,ax = plt.subplots(1,figsize=(12,4))

# plot the tokens
ax.plot(tokens,'ks',markersize=12,markerfacecolor=[.7,.7,.9])
ax.set(xlabel='Character index',yticks=range(len(char_vocab)))
ax.grid(linestyle='--',axis='y')

# invisible axis for right-hand-side labels
ax2 = ax.twinx()
ax2.plot(tokens,alpha=0)
ax2.set(yticks=range(len(char_vocab)),yticklabels=char_vocab)

plt.tight_layout()
plt.show()
```

# Part 2: Word Based Tokenization

## Task 1
- Write code to create a vocab of unique words in the same text that
you worked with in Part 1. Print the vocab and report the number
of tokens in the text and in the vocab.
Visualize the tokenization as in Figure 2.4.

```python
txt ='The way you do anything is the way you do everything.'
words=txt.split()
print("the tokens are:",words)
```

```python
word_vocab=list(sorted(set(words)))
print("the unique vocabs are",word_vocab)
```

```python
len(words),len(word_vocab)
```

```python
print('Token :',end='')
for v in word_vocab:
  print(f' "{v}" | ', end='')
print('\n','-'*131)

print('Index :',end='')
[print(f' {i:5}  | ',end='') for i in range(len(word_vocab))];
```

```python
tokens=[0]*len(words)
for i,c in enumerate(words):
  tokens[i]=word_vocab.index(c)

print(tokens)
```

```python
_,ax = plt.subplots(1,figsize=(12,4))

# plot the tokens
ax.plot(tokens,'ks',markersize=12,markerfacecolor=[.7,.7,.9])
ax.set(xlabel='word index',yticks=range(len(word_vocab)))
ax.grid(linestyle='--',axis='y')

# invisible axis for right-hand-side labels
ax2 = ax.twinx()
ax2.plot(tokens,alpha=0)
ax2.set(yticks=range(len(word_vocab)),yticklabels=word_vocab)

plt.tight_layout()
plt.show()
```

# Part 3:The OpenAi Gpt -2 Tokenizer

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained('gpt2')
```

```python
tokenizer.vocab_size
```

```python
gpt2_tokens=tokenizer.encode(txt)
```

```python
gpt2_tokens = tokenizer.encode(txt)
for token in gpt2_tokens:
  toktext = tokenizer.decode([token])
  print(f'{token:4} is "{toktext}"')
```
